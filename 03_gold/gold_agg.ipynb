```python
# COMMAND ----------
# MAGIC %md
# MAGIC # Phase 1: Gold Aggregation Layer - Analytics Ready

# COMMAND ----------
# Create Gold schema in Unity Catalog
spark.sql("CREATE DATABASE IF NOT EXISTS personal_finance_gold;")
spark.sql("USE personal_finance_gold;")

# COMMAND ----------
# 1. Create Gold TFSA Limits Master Table
df_cra_clean = spark.table("personal_finance_silver.cra_limits_clean")

df_cra_clean.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable("personal_finance_gold.dim_tfsa_limits")

# COMMAND ----------
# 2. Create Gold Interest Rates Master Table
df_boc_clean = spark.table("personal_finance_silver.boc_rates_clean")

df_boc_clean.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable("personal_finance_gold.fct_boc_policy_rates")

print("Gold layer aggregation complete: Master dimension and fact tables are ready for agent tool querying!")
```
